##### 📌 Step 1: Import & Load Dataset

```python
import pandas as pd

# Load dataset with proper encoding
df = pd.read_csv(r"C:\Users\guddi\OneDrive\文档\GOURI\Data Analyst course\Minor project\Mobiles Dataset (2025).csv", encoding="latin1")

# Show first few rows
print(df.head())
```

##### 📌 Step 2: Check Data Info & Missing Value

```python
print(df.info())
print(df.isnull().sum())
```

##### 📌 Step 3: Convert Columns to Proper Data Types

```python
# Remove extra text (like 'GB', 'mAh', etc.) and convert to numeric
# Clean RAM column
def clean_ram(val):
    if pd.isna(val):  # handle NaN
        return None
    val = str(val).replace("GB", "").strip()  # remove "GB"
    if "/" in val:   # if multiple values like "8 / 12"
        nums = [float(x) for x in val.split("/") if x.strip().isdigit()]
        return max(nums) if nums else None  # take the max option
    try:
        return float(val)
    except:
        return None

df['RAM'] = df['RAM'].apply(clean_ram)


# Clean Battery Capacity
df['Battery Capacity'] = (
    df['Battery Capacity']
    .astype(str)                          # make sure all are strings
    .str.replace(",", "", regex=False)    # remove commas
    .str.replace("mAh", "", regex=False)  # remove "mAh"
    .str.strip()                          # remove spaces
)

df['Battery Capacity'] = pd.to_numeric(df['Battery Capacity'], errors="coerce")


# Clean screen size
import re

def clean_screen_size(x):
    if pd.isna(x):
        return None
    # Extract numbers + decimal from string
    match = re.search(r"[\d\.]+", str(x))
    if match:
        return float(match.group())
    return None

df['Screen Size'] = df['Screen Size'].apply(clean_screen_size)



df['Mobile Weight'] = pd.to_numeric(df['Mobile Weight'], errors='coerce')

# Convert prices
price_cols = [
    'Launched Price (Pakistan)',
    'Launched Price (India)',
    'Launched Price (China)',
    'Launched Price (USA)',
    'Launched Price (Dubai)'
]

for col in price_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

#### Step 3: Exploratory Data Analysis (EDA)

```python
# Top companies by number of models
print(df['Company Name'].value_counts().head())

# Average RAM per company
print(df.groupby('Company Name')['RAM'].mean().sort_values(ascending=False).head())

# Correlation between specs
print(df.corr(numeric_only=True))
```

#### 📌 Step 5: Visualization

```python
import matplotlib.pyplot as plt

# Average price (India) by company
avg_price = df.groupby('Company Name')['Launched Price (India)'].mean().sort_values(ascending=False).head(10)

avg_price.plot(kind='bar', figsize=(10,6))
plt.title("Top 10 Companies by Average Launch Price (India)")
plt.ylabel("Price in INR")
plt.show()
```

```python
#Launch price in India

# Clean Launched Price (India) column
df['Launched Price (India)'] = (
    df['Launched Price (India)']
    .astype(str)                                # ensure string
    .str.replace(',', '', regex=False)          # remove commas like 1,20,000
    .str.replace('₹', '', regex=False)          # remove rupee symbol if present
    .str.strip()                                # remove spaces
)

# Convert to numeric (invalids → NaN)
df['Launched Price (India)'] = pd.to_numeric(df['Launched Price (India)'], errors='coerce')

# Now re-run your groupby + plot
avg_price_india = df.groupby("Company Name")["Launched Price (India)"].mean().sort_values(ascending=False).head(10)

avg_price_india.plot(kind='bar', figsize=(10,5), title="Top 10 Companies by Average Launch Price (India)")
plt.ylabel("Price in INR")
plt.show()
```
