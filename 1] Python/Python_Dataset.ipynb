```python
# 📌 Step 1: Import Libraries
import pandas as pd
import numpy as np
```

```python
# Load dataset
file_path = "Mobiles Dataset (2025).csv"
df = pd.read_csv(file_path, encoding="ISO-8859-1")
```

```python
# Preview
df.head()
```

```python
import re

def extract_number(text):
    """
    Extracts number(s) from a string.
    If multiple numbers, return the maximum.
    """
    if pd.isna(text):
        return np.nan
    numbers = re.findall(r"\d+\.?\d*", str(text))  # find numbers
    if not numbers:
        return np.nan
    numbers = [float(n) for n in numbers]
    return max(numbers)   # take max (e.g., "8 / 12" -> 12)

# Apply function to columns
df["RAM"] = df["RAM"].apply(extract_number)
df["Front Camera"] = df["Front Camera"].apply(extract_number)
df["Back Camera"] = df["Back Camera"].apply(extract_number)
df["Mobile Weight"] = df["Mobile Weight"].apply(extract_number)
df["Battery Capacity"] = df["Battery Capacity"].apply(extract_number)
df["Screen Size"] = df["Screen Size"].apply(extract_number)

# Prices cleanup (extract numbers, keep max if multiple)
for col in ["Launched Price (Pakistan)", "Launched Price (India)", 
            "Launched Price (China)", "Launched Price (USA)", "Launched Price (Dubai)"]:
    df[col] = df[col].apply(extract_number)

df.info()
df.head()
```

```python
import pandas as pd

# Load raw dataset
file_path = "/mnt/data/Mobiles Dataset (2025).csv"
mobiles_df = pd.read_csv(file_path)

# Cleaning
mobiles_df = mobiles_df.drop_duplicates()
mobiles_df = mobiles_df.dropna(how="all")
mobiles_df.columns = mobiles_df.columns.str.strip().str.replace(" ", "_")

# Save cleaned dataset
cleaned_file = "/mnt/data/Mobiles_Dataset_Cleaned.csv"
mobiles_df.to_csv(cleaned_file, index=False)

print("✅ Cleaned dataset saved at:", cleaned_file)
```

```python
# 📌 Step 2: Data Cleaning (safe version)

# Convert everything to string first before cleaning
df["Mobile Weight"] = (
    df["Mobile Weight"].astype(str).str.replace("g", "", regex=False).replace("nan", np.nan).astype(float)
)

df["RAM"] = (
    df["RAM"].astype(str).str.replace("GB", "", regex=False).replace("nan", np.nan).astype(float)
)

df["Front Camera"] = (
    df["Front Camera"].astype(str).str.replace("MP", "", regex=False).replace("nan", np.nan).astype(float)
)

df["Back Camera"] = (
    df["Back Camera"].astype(str).str.replace("MP", "", regex=False).replace("nan", np.nan).astype(float)
)

df["Battery Capacity"] = (
    df["Battery Capacity"].astype(str)
    .str.replace("mAh", "", regex=False)
    .str.replace(",", "", regex=False)
    .replace("nan", np.nan)
    .astype(float)
)

df["Screen Size"] = (
    df["Screen Size"].astype(str).str.replace("inches", "", regex=False).replace("nan", np.nan).astype(float)
)

# Prices cleanup
for col in ["Launched Price (Pakistan)", "Launched Price (India)", 
            "Launched Price (China)", "Launched Price (USA)", "Launched Price (Dubai)"]:
    df[col] = (
        df[col].astype(str)
        .str.replace(r"[^\d.]", "", regex=True)   # keep only digits & decimals
        .replace("", np.nan)
        .astype(float)
    )

df.info()
df.head()
```
